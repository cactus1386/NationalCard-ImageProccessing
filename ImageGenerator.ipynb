```python
from PIL import ImageDraw, ImageFont, Image
import pandas as pd
import os
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
def generate(text):
  image_size = (200, 70)

  bg = Image.open("/content/drive/MyDrive/ImageGenerate/bg.jpg")
  draw = ImageDraw.Draw(bg)


  font = ImageFont.truetype('/content/drive/MyDrive/ImageGenerate/iransansdn.ttf', 40)

  bbox = draw.textbbox((0, 0), text, font=font)
  text_width = bbox[2] - bbox[0]
  text_height = bbox[3] - bbox[1]

  position = ((image_size[0] - text_width) // 2, ((image_size[1] - text_height) // 2) - 17)

  draw.text(position, text, font=font, fill=(0, 0, 0))

  bg.save(f'/content/drive/MyDrive/ImageGenerate/{text}.jpg')
```

```python
generate('امیرعلی')
```
