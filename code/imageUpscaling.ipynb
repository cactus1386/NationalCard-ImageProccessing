```python
import cv2
import numpy as np
from skimage import io, transform
from PIL import Image
from google.colab.patches import cv2_imshow
```

```python
path = '/content/drive/MyDrive/images/1.jpg'
img = cv2.imread(path)
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
scale = 2
upscaled = cv2.resize(gray, None, fx=scale, fy=scale, interpolation=cv2.INTER_CUBIC)
contrast = cv2.equalizeHist(upscaled)
```

```python
brightness_reduction = -20
adjusted_brightness = cv2.convertScaleAbs(contrast, beta=brightness_reduction)
```

```python
cv2_imshow(adjusted_brightness)
```
