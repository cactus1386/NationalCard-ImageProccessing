```python
from PIL import ImageDraw, ImageFont, Image, ImageEnhance, ImageFilter
import pandas as pd
import os
import random
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
def generate(text, bg):
    image_size = (200, 70)

    # بارگذاری بک‌گراند
    bg = Image.open(f"/content/drive/MyDrive/ImageGenerate/{bg}.jpg")
    bg = bg.resize(image_size, Image.Resampling.LANCZOS)  # تغییر اندازه بک‌گراند
    draw = ImageDraw.Draw(bg)

    # بارگذاری فونت
    font = ImageFont.truetype('/content/drive/MyDrive/ImageGenerate/iransansdn.ttf', 40)

    # محاسبه اندازه متن
    bbox = draw.textbbox((0, 0), text, font=font)
    text_width = bbox[2] - bbox[0]
    text_height = bbox[3] - bbox[1]

    # محاسبه موقعیت متن (کمی بالاتر)
    position = ((image_size[0] - text_width) // 2, ((image_size[1] - text_height) // 2) - 17)

    # نوشتن متن روی تصویر اصلی
    draw.text(position, text, font=font, fill=(0, 0, 0))

    # حذف اسلش‌ها از نام فایل
    t = text.replace('/', '')

    # ذخیره تصویر اصلی
    output_path = f'/content/drive/MyDrive/ImageGenerate/images/{t}.jpg'
    bg.save(output_path)

    # کپی 1: تغییر روشنایی (Brightness)
    bg_bright = ImageEnhance.Brightness(bg).enhance(1.5)  # روشن‌تر کردن (1.5 برابر)
    output_path_bright = f'/content/drive/MyDrive/ImageGenerate/images/{t}b.jpg'
    bg_bright.save(output_path_bright)

    # کپی 2: تار کردن (Blur)
    bg_blur = bg.filter(ImageFilter.GaussianBlur(radius=1.4))  # تار کردن با شعاع 1
    output_path_blur = f'/content/drive/MyDrive/ImageGenerate/images/{t}t.jpg'
    bg_blur.save(output_path_blur)
```

```python
df = pd.read_csv('/content/drive/MyDrive/ImageGenerate/iranian_names.csv')

for index, row in df.iterrows():
  text = row["Name"]
  generate(text, 'bg_name')
```

```python
def to_persian_digits(number):
    persian_digits = {'0': '۰', '1': '۱', '2': '۲', '3': '۳', '4': '۴',
                      '5': '۵', '6': '۶', '7': '۷', '8': '۸', '9': '۹'}
    return ''.join(persian_digits.get(digit, digit) for digit in str(number))
```

```python
def random_date():
    year = random.randint(1300, 1450)
    month = random.randint(1, 12)
    day = random.randint(1, 30)
    year_str = to_persian_digits(f"{year:04d}")
    month_str = to_persian_digits(f"{month:02d}")
    day_str = to_persian_digits(f"{day:02d}")
    return f"{year_str}/{month_str}/{day_str}"

dates1 = [random_date() for _ in range(400)]
dates2 = [random_date() for _ in range(400)]

df1 = pd.DataFrame(dates1, columns=['Birth'])
df2 = pd.DataFrame(dates2, columns=['Date'])

df1.to_csv('/content/drive/MyDrive/ImageGenerate/Birth.csv', index=False, encoding='utf-8-sig')
df2.to_csv('/content/drive/MyDrive/ImageGenerate/Date.csv', index=False, encoding='utf-8-sig')
```
